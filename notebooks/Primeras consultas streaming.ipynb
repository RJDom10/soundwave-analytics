```python
from pyspark.sql import SparkSession
import pyspark.sql.functions as F

spark = SparkSession.builder \
    .appName("Soundwave-Analytics-Exploration") \
    .master("local[*]") \
    .config("spark.jars.packages", "org.apache.spark:spark-sql-kafka-0-10_2.12:3.5.0") \
    .getOrCreate()
```

```python
path_bronze = "hdfs://namenode:9000/data/bronze/play_events"
df_bronze = spark.read.parquet(path_bronze)

# Esto debería imprimir las primeras 5 filas en formato texto casi de inmediato
df_bronze.show(5, truncate=False)
```

```python
# Tomamos los primeros 10 registros y los pasamos al formato visual de Pandas
df_bronze.limit(10).toPandas()
```

```python
# 1. Filtramos solo los eventos donde el usuario dio 'like'
# 2. Agrupamos por el ID de la canción
# 3. Contamos cuántos likes tiene cada una
# 4. Ordenamos de mayor a menor y nos quedamos con las 3 mejores

df_bronze = spark.read.parquet("hdfs://namenode:9000/data/bronze/play_events")
df_top_songs = df_bronze \
    .filter(F.col("event_type") == "like") \
    .groupBy("track_id") \
    .count() \
    .withColumnRenamed("count", "total_likes") \
    .orderBy("total_likes", ascending=False) \
    .limit(3)

# Lo pasamos a Pandas para que Jupyter lo pinte como tabla elegante
df_top_songs.toPandas()
```
